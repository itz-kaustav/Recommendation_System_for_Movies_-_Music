```python
models={
    "SVC1":SVC(kernel='linear'),
    "SVC2":SVC(kernel="rbf"),
    "LOGISTIC":LogisticRegression(),
    "DC":DecisionTreeClassifier(random_state=5050),
    "RF":RandomForestClassifier(n_estimators=200,random_state=2020),
    "knc":KNeighborsClassifier(n_neighbors=5),
    "NB":GaussianNB()
}
result={}
```

```python
import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler , LabelEncoder
import joblib
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score,f1_score


categorical_cols = x_train.select_dtypes(include='object').columns


x_train_encoded = x_train.copy()
x_test_encoded = x_test.copy()

for col in categorical_cols:
    le = LabelEncoder()

    all_unique_values = pd.concat([x_train_encoded[col], x_test_encoded[col]]).unique()
    le.fit(all_unique_values)

    x_train_encoded[col] = le.transform(x_train_encoded[col])
    x_test_encoded[col] = le.transform(x_test_encoded[col])

for name,model in models.items():

  pipe=Pipeline([
      ("scale",MinMaxScaler()),
      ("model",model)
  ])


  pipe.fit(x_train_encoded,y_train)
  y_pred=pipe.predict(x_test_encoded)

  acc=accuracy_score(y_test,y_pred)

  pre=precision_score(y_test,y_pred, average='weighted', zero_division=0)
  f1=f1_score(y_test,y_pred, average='weighted', zero_division=0)
  rec=recall_score(y_test,y_pred, average='weighted', zero_division=0)
  cm=confusion_matrix(y_test,y_pred)

  result[name]={
      "model":pipe,
      "acc":acc,
      "pre":pre,
      "f1":f1,
      "rec":rec,
      "cm":cm
  }
  print("\n",name)
  print("acc",acc),
  print("pre",pre),
  print("rc",rec),
  print("f1",f1),
  print("cm",cm)
```

```python
best_model=max(result,key=lambda x:result[x]["acc"])
best_pipe=result[best_model]["model"]
```
